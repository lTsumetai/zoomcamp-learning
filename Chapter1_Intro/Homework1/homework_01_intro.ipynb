```python
# Maximum Fuel Efficiency for Asian Cars
df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
```

```python
# Finding median value of horsepower
median_before = df['horsepower'].median()
median_before
```

```python
most_frequent = df['horsepower'].mode()[0]
most_frequent
```

```python
df['horsepower'] = df['horsepower'].fillna(most_frequent)
```

```python
median_after = df['horsepower'].median()
median_after
```

```python
print(f'Before: {median_before}, After: {median_after}')
if median_after > median_before:
    print('Yes, it increased')
elif median_after < median_before:
    print('Yes, it decreased')
else:
    print('No')
```

```python
# Sum of Weight of Cars from Asia
asia = df[df['origin'] == 'Asia']
subset = asia[['vehicle_weight', 'model_year']].head(7)
subset
```

```python
X = subset.values
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
w
```
